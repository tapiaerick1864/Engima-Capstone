```python
import numpy as np

# Dictionary to map digits to their spelled-out forms
digit_map = {
    '0': 'zero', '1': 'one', '2': 'two', '3': 'three', '4': 'four',
    '5': 'five', '6': 'six', '7': 'seven', '8': 'eight', '9': 'nine'
}

# Input/Output filenames
input_filename = r"C:\Users\tapia\Desktop\Python stuff\Classes\Capstone\training data\bee movie script.txt"
output_filename = 'Processed Bee.txt'

print("="*70)
print("BEE MOVIE SCRIPT TEXT PROCESSOR")
print("="*70)

# Load the text file
print(f"\nReading input file: {input_filename}")
with open(input_filename, 'r') as file:
    text = file.read()

print(f"Original text length: {len(text)} characters")

# Replace all digit characters with their spelled-out forms
print("\nProcessing: Converting digits to words...")
processed_text = ''
for char in text:
    if char.isdigit():
        processed_text += digit_map[char]
    else:
        processed_text += char

# Remove punctuation and spaces - keep only alphanumeric characters
print("Processing: Removing punctuation and spaces...")
cleaned_text = ''.join(char for char in processed_text if char.isalnum())

print(f"Processed text length: {len(cleaned_text)} characters")

# Convert to uppercase for Enigma compatibility
cleaned_text = cleaned_text.upper()

# Display preview
print("\n" + "-"*70)
print("PREVIEW (first 200 characters):")
print("-"*70)
print(cleaned_text[:200])
print("\n" + "-"*70)
print("PREVIEW (last 100 characters):")
print("-"*70)
print(cleaned_text[-100:])
print("-"*70)

# Save as CONTINUOUS TEXT (not matrix format!)
print(f"\nSaving to: {output_filename}")
with open(output_filename, 'w') as file:
    file.write(cleaned_text)

print(f"✓ Successfully saved {len(cleaned_text)} characters")

# Optional: Also create a numpy array for ML purposes if needed
# But save it separately, not as the main output
print("\nCreating numpy matrix for ML analysis...")
char_array = np.array(list(cleaned_text))
print(f"Numpy array shape: {char_array.shape}")

# Save numpy version separately (for ML training)
ml_output_filename = 'bee_movie_matrix.npy'
np.save(ml_output_filename, char_array)
print(f"✓ Numpy array saved to: {ml_output_filename}")

print("\n" + "="*70)
print("PROCESSING COMPLETE!")
print("="*70)
print(f"Text file (for Enigma/Bombe): {output_filename}")
print(f"Numpy file (for ML training): {ml_output_filename}")
print("="*70)
```
